```python

import os
import json
import torch
from PIL import Image
from tqdm.auto import tqdm

DATA_DIR = "/kaggle/working/data"
CACHE_DIR = "/kaggle/working/clip_cache"

BATCH_SIZE = 8
SHARD_SIZE = 250
MAX_LENGTH = 77

os.makedirs(CACHE_DIR, exist_ok=True)

def extract_and_cache(split_name, jsonl_file):
    split_cache = os.path.join(CACHE_DIR, split_name)
    os.makedirs(split_cache, exist_ok=True)

    with open(jsonl_file, "r", encoding="utf-8") as f:
        records = [json.loads(line) for line in f if line.strip()]

    model.eval()
    print(f"\nExtracting {split_name}: {len(records)} samples")

    for start in tqdm(range(0, len(records), SHARD_SIZE)):
        end = min(start + SHARD_SIZE, len(records))

        shard_path = os.path.join(
            split_cache, f"part_{start // SHARD_SIZE:04d}.pt"
        )

        if os.path.exists(shard_path):
            print("Skipping existing shard:", shard_path)
            continue

        shard_records = records[start:end]

        shard_data = {
            "image_pooled": [],
            "text_pooled": [],
            "image_tokens": [],
            "text_tokens": [],
            "attention_mask": [],
            "input_ids": [],
            "labels": [],
            "ids": [],
            "texts": []
        }

        for batch_start in range(0, len(shard_records), BATCH_SIZE):
            batch_records = shard_records[
                batch_start:batch_start + BATCH_SIZE
            ]

            images = []

            for record in batch_records:
                image_path = os.path.join(DATA_DIR, record["img"])

                if not os.path.isfile(image_path):
                    raise FileNotFoundError(image_path)

                with Image.open(image_path) as image:
                    images.append(image.convert("RGB"))

            texts = [record["text"] for record in batch_records]

            inputs = processor(
                text=texts,
                images=images,
                return_tensors="pt",
                padding="max_length",
                truncation=True,
                max_length=MAX_LENGTH
            )

            inputs = {
                key: value.to("cuda")
                for key, value in inputs.items()
            }

            with torch.inference_mode():
                with torch.autocast(
                    device_type="cuda",
                    dtype=torch.float16
                ):
                    outputs = model(**inputs, return_dict=True)

            shard_data["image_pooled"].append(
                outputs.image_embeds.cpu().half()
            )
            shard_data["text_pooled"].append(
                outputs.text_embeds.cpu().half()
            )
            shard_data["image_tokens"].append(
                outputs.vision_model_output.last_hidden_state.cpu().half()
            )
            shard_data["text_tokens"].append(
                outputs.text_model_output.last_hidden_state.cpu().half()
            )
            shard_data["attention_mask"].append(
                inputs["attention_mask"].cpu().bool()
            )
            shard_data["input_ids"].append(
                inputs["input_ids"].cpu()
            )

            shard_data["labels"].extend(
                [int(record["label"]) for record in batch_records]
            )
            shard_data["ids"].extend(
                [record.get("id") for record in batch_records]
            )
            shard_data["texts"].extend(texts)

        for key in [
            "image_pooled", "text_pooled",
            "image_tokens", "text_tokens",
            "attention_mask", "input_ids"
        ]:
            shard_data[key] = torch.cat(shard_data[key], dim=0)

        shard_data["labels"] = torch.tensor(
            shard_data["labels"], dtype=torch.long
        )

        torch.save(shard_data, shard_path)

        del shard_data
        torch.cuda.empty_cache()

    print(f"{split_name} extraction completed!")


extract_and_cache(
    split_name="dev",
    jsonl_file=os.path.join(DATA_DIR, "dev_yolo.jsonl")
)
```

```python
import os
import sys
import json
import torch
import pandas as pd

# Import the exact architecture
sys.path.insert(
    0,
    "/kaggle/input/datasets/aarnarajput/irisai-final-research-handoff"
)
from advanced_fusion import BilinearFusionHead

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

CHECKPOINT_DIR = (
    "/kaggle/input/datasets/aarnarajput/"
    "irisai-final-research-handoff/irisAI_experiments/"
    "bilinear_originals"
)

CACHE_DIR = "/kaggle/working/clip_cache/dev"

# Load the two dev shards in their original order
shards = []
for filename in ["part_0000.pt", "part_0001.pt"]:
    path = os.path.join(CACHE_DIR, filename)
    shards.append(
        torch.load(path, map_location="cpu", weights_only=False)
    )

image_pooled = torch.cat(
    [s["image_pooled"] for s in shards], dim=0
)
text_pooled = torch.cat(
    [s["text_pooled"] for s in shards], dim=0
)

ids = sum([s["ids"] for s in shards], [])
labels = torch.cat([s["labels"] for s in shards], dim=0)

# First 10 samples only
images = image_pooled[:10].to(DEVICE).float()
texts = text_pooled[:10].to(DEVICE).float()

results = {}

for seed in [7, 42, 123]:

    checkpoint_path = os.path.join(
        CHECKPOINT_DIR, f"bilinear_seed_{seed}.pth"
    )

    checkpoint = torch.load(
        checkpoint_path,
        map_location=DEVICE,
        weights_only=False
    )

    model = BilinearFusionHead(
        image_dim=768,
        text_dim=768,
        fusion_dim=512,
        hidden_dim=256,
        dropout=0.3
    ).to(DEVICE)

    model.load_state_dict(checkpoint["model_state_dict"])
    model.eval()

    with torch.inference_mode():
        logits = model(images, texts)
        probabilities = torch.sigmoid(logits)

    results[f"probability_seed_{seed}"] = (
        probabilities.cpu().numpy()
    )

    print(
        f"Seed {seed} | Epoch: {checkpoint['epoch']} | "
        f"Validation Macro-F1: {checkpoint['validation_macro_f1']}"
    )

# Create output table
df = pd.DataFrame({
    "id": ids[:10],
    "label": labels[:10].tolist(),
    **results
})

print("\nFirst 10 dev probabilities:")
display(df)

# Save CSV
output_path = (
    "/kaggle/working/bilinear_first10_probabilities.csv"
)
df.to_csv(output_path, index=False)

print("\nSaved:", output_path)
```

```python
import os
import json
import torch
import zipfile
import inspect
import shutil

WORK = "/kaggle/working"
CACHE_DIR = f"{WORK}/clip_cache/dev"
DEV_JSONL = f"{WORK}/data/dev_yolo.jsonl"

# 1. Load original dev records
with open(DEV_JSONL, "r", encoding="utf-8") as f:
    records = [json.loads(line) for line in f if line.strip()]

# 2. Load both feature shards in original order
shards = []
for filename in ["part_0000.pt", "part_0001.pt"]:
    shards.append(
        torch.load(
            os.path.join(CACHE_DIR, filename),
            map_location="cpu",
            weights_only=False
        )
    )

# 3. Combine features
image_pooled = torch.cat(
    [s["image_pooled"] for s in shards], dim=0
)
text_pooled = torch.cat(
    [s["text_pooled"] for s in shards], dim=0
)
labels = torch.cat([s["labels"] for s in shards], dim=0)
ids = sum([s["ids"] for s in shards], [])
texts = sum([s["texts"] for s in shards], [])

# 4. Verify exact order, IDs, labels and captions
assert len(records) == 500
assert ids == [r["id"] for r in records]
assert labels.tolist() == [r["label"] for r in records]
assert texts == [r["text"] for r in records]
assert image_pooled.shape == (500, 768)
assert text_pooled.shape == (500, 768)

print("All 500 records verified.")
print("Image features:", image_pooled.shape)
print("Text features:", text_pooled.shape)

# 5. Save compact feature cache
feature_path = f"{WORK}/dev_clip_features.pt"

torch.save({
    "image_pooled": image_pooled,
    "text_pooled": text_pooled,
    "labels": labels,
    "texts": texts,
    "ids": ids
}, feature_path)

# 6. Save the exact extraction function from the notebook
function_source = inspect.getsource(extract_and_cache)

function_path = f"{WORK}/extract_and_cache_function.py"

with open(function_path, "w", encoding="utf-8") as f:
    f.write(function_source)

# 7. Check probability CSV exists
probability_path = (
    f"{WORK}/bilinear_first10_probabilities.csv"
)
assert os.path.exists(probability_path), (
    "Probability CSV not found. Run Step 13 first."
)

# 8. Create README
readme = """
irisAI - Dev Feature and Bilinear Inference Handoff

Contents:
1. dev_clip_features.pt
   - 500 dev samples, original dev order
   - image_pooled: [500, 768]
   - text_pooled: [500, 768]
   - labels: [500]
   - texts: exact CLIP captions, including detected objects
   - ids: original dev IDs

2. extract_and_cache_function.py
   - Exact extract_and_cache function from notebook cell 14.

3. bilinear_first10_probabilities.csv
   - Class-1 probabilities for the first 10 dev samples.
   - Separate predictions from seeds 7, 42 and 123.
   - Probabilities are sigmoid of raw logits, without temperature calibration.

Feature extraction:
- Backbone: openai/clip-vit-large-patch14
- Pooled embedding dimensions: 768 each
- Captions sourced from dev_yolo.jsonl
- Original dev ordering preserved.

Bilinear checkpoints:
- bilinear_seed_7.pth
- bilinear_seed_42.pth
- bilinear_seed_123.pth

No model retraining was performed for this handoff.
"""

readme_path = f"{WORK}/README_friend_handoff.txt"

with open(readme_path, "w", encoding="utf-8") as f:
    f.write(readme.strip())

# 9. Zip all deliverables
zip_path = f"{WORK}/irisAI_friend_handoff.zip"

with zipfile.ZipFile(
    zip_path, "w", zipfile.ZIP_DEFLATED
) as z:
    z.write(feature_path, "dev_clip_features.pt")
    z.write(function_path, "extract_and_cache_function.py")
    z.write(probability_path, "bilinear_first10_probabilities.csv")
    z.write(readme_path, "README_friend_handoff.txt")

print("\nHandoff ZIP created:", zip_path)
print("ZIP size (MB):", round(os.path.getsize(zip_path) / (1024**2), 2))
print("\nZIP contents:")
with zipfile.ZipFile(zip_path, "r") as z:
    for name in z.namelist():
        print("-", name)
```
